```python
#packages
# we import functions using this:
import numpy # numpy is a common package for handling numeric data
```

```python
## reading data with numpy

numpy.loadtxt(fname= 'molerat_activity_v2.csv',
              delimiter=',')

```

```python
data= numpy.loadtxt(fname='molerat_activity_v2.csv',
                    delimiter=',')
```

```python
print(data)
```

```python
print(type(data)) # this prints the type of the data object
```

```python
#we can also print the type of the data in the columns
print(data.dtype)
#and shape to confirm we have all the data
print(data.shape)
```

```python
### slicing using [  ]
data[0,0] # show the data in the first row and first column
data[0,1] # show the data in the first row and second column
print('the value in the first row, second column is', data[0,1])
```

```python
# while rows
data[0,] # first row

## whole columns
data[:,0] # first column, all rows is specified by a : with nothing on either side
```

```python
#slicing ranges
print(data[0:10,]) # first 10 rows, all columns, print row 0 to row 10
print(data[:, 0:10]) # first collumns all rows, all rows need :
#important note: the first value is INCLUSIVE, the second value is EXCLUSIVE
# 0:10 will print from row/column 0 to row/column 9.
```

```python
## these data are naked mole-rat activity
## rows are individuals
## columns are observation sessions
## cells are the count of how many times an individual was active in each session
## what is the mean number of activities observed per session

print(numpy.mean(data))
```

```python
## what about individual mole-rats?

numpy.mean(data[0,]) # the first individual (in the first row)
print('activity from individual 1:', numpy.mean(data[0,]))
```

```python
#what about sessions?

numpy.mean(data[:,0]) # mean for observation session 1
```

```python
import matplotlib.pyplot
image = matplotlib.pyplot.imshow(data, aspect='auto')
matplotlib.pyplot.show()
```

```python
ave_day = numpy.mean(data, axis=0)
ave_day_plot = matplotlib.pyplot.plot(ave_day)
matplotlib.pyplot.show
```

```python
ave_MR = numpy.mean(data, axis=1)
ave_MR_plot = matplotlib.pyplot.plot(ave_MR)
matplotlib.pyplot.show()
```

```python
min_plot = matplotlib.pyplot.plot(numpy.min(data, axis=1))
matplotlib.pyplot.show()
```

```python
fig = matplotlib.pyplot.figure(figsize=(10.0, 3.0)) # size of the figure

axis1 = fig.add_subplot(1,3,1)
axis2 = fig.add_subplot(1,3,2)
axis3 = fig.add_subplot(1,3,3)

axis1.set_ylabel('average')
axis1.plot(numpy.mean(data, axis=1))

axis2.set_ylabel('max')
axis2.plot(numpy.max(data, axis=1))

axis3.set_ylabel('min')
axis3.plot(numpy.min(data, axis=1))

fig.tight_layout()

matplotlib.pyplot.savefig('activity.png')
matplotlib.pyplot.show
```
